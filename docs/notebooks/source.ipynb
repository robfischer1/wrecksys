# Chapter 1: Source Data

This capstone project (henceforth *wrecksys*) is a sequential recommendation system based on a subset of the Goodreads Book Graph dataset. The Goodreads dataset is publicly available and was originally compiled for use in __[Item Recommendation on Monotonic Behavior Chains](https://mengtingwan.github.io/paper/recsys18_mwan.pdf)__ and __[Fine-Grained Spoiler Detection from Large-Scale Review Corpora](https://mengtingwan.github.io/paper/acl19_mwan.pdf)__

The full dataset is available for reference __[on GitHub](https://mengtingwan.github.io/data/goodreads)__


## Goodreads Book Graph

The full dataset, compiled in 2017, documents interactions between 876,145 users and 2.3 million books on [Goodreads](http://www.goodreads.com). The resulting 228 million interactions are available for download as compressed JSON files. The authors have also compiled smaller subsets of these interactions, divided by genre.

Wrecksys is built on the Fantasy & Paranormal subset, which contains a miniscule 55.3 million interactions. This subset consists of three files, which have been nicknamed for readability:
* **books** Metadata for each of the 258,585 books which appear in the dataset.
* **ratings** Contains details on each of the 55.3 million user-book interactions.
* **reviews** The full text of user reviews.

Two additional source files are included from the complete dataset:
* **authors** Contains additional details for each of the 829,000 authors appearing in the full dataset.
* **works** A format and edition independent representation of a Goodreads book.

## Imports and Introduction

Even this small subset of the Goodreads Dataset is quite large: It contains 10GB of .gzip archives which decompress into approximately 30GB of JSON files. Working with large JSON files presents a number of challenges, as does including them for submission. This notebook attempts to make the process as painless as possible:

- GoodreadsData, imported below, creates an abstraction layer over the source files. 
    - Individual files are accessible by description (books, authors, etc.) for clarity.
    - Single examples are generated for each file to examine the structure of the underlying data before downloading or loading.
    - pyArrow is used to read the source data and convert fields to the appropriate data types before saving. 
    - The dataframe() method accepts a list of columns to avoid loading un-needed data into memory.
- Many steps in Data Analysis and Data Preparation overlap: the next section executes both in parallel.
    - When training a model locally, these steps are completed automatically the first time the dataset is accessed.
- The original dataset is downloaded entirely to comply with the capstone requirements, but converted from JSON to Feather in order to reduce performance bottlenecks. Individual fields are cast to appropriate dtypes during conversion. 
    - Only the necessary columns are loaded during data exploration due to the size of the underlying files.  
    - The full source data is preserved in /content/goodreads-recs/data if you'd like to explore on your own.
